```python
#Task 1: Database Design and Table Creation (35 points)
import sqlite3

# Step 1: Create database connection
conn = sqlite3.connect("bookstore.db")
cursor = conn.cursor()
print("✓ Database connection established")

# Step 2: Write SQL CREATE TABLE statements
create_books_table = """
CREATE TABLE IF NOT EXISTS Books (
    book_id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT NOT NULL,
    author TEXT NOT NULL,
    price REAL NOT NULL,
    stock_quantity INTEGER DEFAULT 0
);
"""

create_customers_table = """
CREATE TABLE IF NOT EXISTS Customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    city TEXT,
    join_date TEXT
);
"""

create_orders_table = """
CREATE TABLE IF NOT EXISTS Orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER,
    book_id INTEGER,
    quantity INTEGER NOT NULL,
    order_date TEXT NOT NULL,
    total_amount REAL,
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id),
    FOREIGN KEY (book_id) REFERENCES Books(book_id)
);
"""

# Step 3: Create all three tables
cursor.execute(create_books_table)
print("✓ Books table created successfully")
cursor.execute(create_customers_table)
print("✓ Customers table created successfully")
cursor.execute(create_orders_table)
print("✓ Orders table created successfully")

# Step 4: Display schema of each table using PRAGMA


def show_schema(table_name):
    print(f"\nSchema for {table_name} table:")
    print("cid | name           | type    | notnull | dflt_value | pk")
    print("-" * 60)
    cursor.execute(f"PRAGMA table_info({table_name});")#PRAGMA table_info to fetch schema details and prints them neatly aligned.
    for row in cursor.fetchall():
        # row = (cid, name, type, notnull, dflt_value, pk)
        print(f"{row[0]:<3} | {row[1]:<14} | {row[2]:<7} | {row[3]:<7} | {str(row[4]):<10} | {row[5]}")

show_schema("Books")
show_schema("Customers")
show_schema("Orders")

# Close connection
conn.commit()
conn.close()





#- Connects to (or creates) bookstore.db.
#- Defines SQL statements with proper constraints (primary keys, foreign keys, defaults, uniqueness).
#- Executes the table creation commands.
# Formats the schema output into a tabular style with column headers
#PRAGMA table_info to fetch schema details and prints them neatly aligned.
#- Closes the connection at the end.

```

```python
#Task 2: Data Insertion and Querying (35 points)
import sqlite3

# Connect to database
conn = sqlite3.connect("bookstore.db")
cursor = conn.cursor()
# --- Create Tables ---
cursor.execute("""
CREATE TABLE IF NOT EXISTS Books (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT NOT NULL,
    author TEXT NOT NULL,
    price REAL NOT NULL,
    stock INTEGER NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    city TEXT NOT NULL,
    join_date DATE NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Orders (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    book_id INTEGER NOT NULL,
    quantity INTEGER NOT NULL,
    order_date DATE NOT NULL,
    amount REAL NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES Customers(id),
    FOREIGN KEY (book_id) REFERENCES Books(id)
)
""")


# --- Insert Data ---
books_data = [
    ('Python Programming', 'John Smith', 599.99, 25),
    ('Data Science Handbook', 'Jane Doe', 899.50, 15),
    ('Machine Learning Basics', 'Alan Turing', 1299.00, 10),
    ('SQL Essentials', 'Edgar Codd', 499.99, 30),
    ('Web Development', 'Tim Berners', 799.00, 20)
]

customers_data = [
    ('Rahul Sharma', 'rahul@email.com', 'Mumbai', '2024-01-15'),
    ('Priya Patel', 'priya@email.com', 'Delhi', '2024-02-01'),
    ('Amit Kumar', 'amit@email.com', 'Bangalore', '2024-02-01'),
    ('Sneha Reddy', 'sneha@email.com', 'Hyderabad', '2024-02-10'),
    ('Vikram Singh', 'vikram@email.com', 'Mumbai', '2024-02-15')
]

orders_data = [
    (1, 1, 2, '2024-03-01', 1199.00),
    (1, 2, 1, '2024-03-02', 899.50),
    (2, 1, 1, '2024-03-03', 599.99),
    (2, 3, 1, '2024-03-05', 1299.00),
    (3, 4, 3, '2024-03-07', 1499.97),
    (4, 2, 1, '2024-03-10', 899.50),
    (5, 5, 2, '2024-03-12', 1598.00)
]

cursor.executemany("INSERT INTO Books (book_id,title, author, price, stock_quantity) VALUES (?, ?, ?, ?,?)", books_data)
cursor.executemany("INSERT INTO Customers (customer_id,name, email, city, join_date) VALUES (?, ?, ?, ?,?)", customers_data)
cursor.executemany("INSERT INTO Orders (customer_id,order_id, book_id, quantity, order_date, total_amount) VALUES (?, ?, ?, ?, ?,?)", orders_data)

conn.commit()

# --- Task 2 Outputs ---
print("=== TASK 2: Data Insertion and Querying ===")
print("✓ Inserted 5 books")
print("✓ Inserted 5 customers")
print("✓ Inserted 7 orders")

print("\nAll Books:")
for row in cursor.execute("SELECT id, title, author, price, stock FROM Books"):
    print(row)

print("\nCustomers from Mumbai:")
for row in cursor.execute("SELECT id, name, email, city FROM Customers WHERE city = 'Mumbai'"):
    print(row)

print("\nBooks priced > 800 with stock > 10:")
for row in cursor.execute("SELECT id, title, price, stock FROM Books WHERE price > 800 AND stock > 10"):
    print(row)

print("\nTotal Orders:")
cursor.execute("SELECT COUNT(*) FROM Orders")
print(cursor.fetchone()[0])

print("\nTotal Revenue:")
cursor.execute("SELECT SUM(amount) FROM Orders")
print(cursor.fetchone()[0])

# Close connection
conn.close()

#- Inserted 5 books, 5 customers, 7 orders
#- Customers from Mumbai → Rahul Sharma, Vikram Singh
#- Books priced > 800 and stock > 10 → Data Science Handbook (₹899.50, stock 15)
#- Total Orders → 7
#- Total Revenue → ₹8795.96


```
